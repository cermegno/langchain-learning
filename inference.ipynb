Import libraries and Nvidia key

````python
from langchain_nvidia_ai_endpoints import ChatNVIDIA
from langchain_core.prompts import ChatPromptTemplate
````

````python
from google.colab import userdata
NVIDIA_API_KEY = userdata.get('apikey')
````

Initializing the model

````python
llm = ChatNVIDIA(
    model="meta/llama3-8b-instruct",
    api_key = NVIDIA_API_KEY,
    temperature = 0.5,
    max_completion_tokens = 1024
    )
````

Create a prompt to use with the chain

````python
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful AI assistant."),
    ("user", "{input}")
    ])
````

Create the chain. This syntax replaces the old ```LLMchain()```

````python
chain = prompt | llm
````

Test the chain with "invoke" method

````python
response = chain.invoke({"input": "what is the colour of the sky?"})
print(response.content)
````
